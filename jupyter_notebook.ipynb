```python
import pandas as pd
import sqlite3

nome_csv = 'G51_data.csv'
nome_db = 'G51.db'

df = pd.read_csv(nome_csv)
conn = sqlite3.connect(nome_db)


df[['company_id', 'company_name', 'company_creation_date']].drop_duplicates().to_sql('Company', conn, if_exists='append', index=False)
df[['grid_id', 'grid_name', 'grid_address']].drop_duplicates().to_sql('Grid', conn, if_exists='append', index=False)
df[['plant_id', 'plant_comments', 'company_id']].drop_duplicates().to_sql('Plant', conn, if_exists='append', index=False)
df[['distribution_date', 'energy_supplied_kwh', 'plant_id', 'grid_id']].to_sql('Distribution', conn, if_exists='append', index=False)

conn.close()


```
